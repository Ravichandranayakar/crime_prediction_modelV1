```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load annotations
import json
with open(r"C:\Users\Ravichandran\OneDrive\Desktop\FIR_details.json") as f:
    data = json.load(f)
```

```python
# group by imgae_id
from collections import defaultdict
grouped = defaultdict(list)

for row in data:
    grouped[row['image_id']].append(row)
    
```

```python
records = []

for image_id, rows in grouped.items():
    texts  = [r['text'] for r in rows]
    ocr_text = " ".join(texts)
    ipc = [r['text'] for r in rows if r["category_id"] == 2]
    ps = [r['text'] for r in rows if r['category_id'] == 0]
    year= [r['text'] for r in rows if r['category_id'] == 1]

    records.append({
        'image_id': image_id,
        'ocr_text':ocr_text,
        'ipc_sections': ipc,
        'police_station':ps[0] if ps else None,
        'year':year[0] if year else None,
        'source':'FIR_ICDAR',
})
```

```python
crime = pd.DataFrame(records)
```
